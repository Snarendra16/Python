```python
# Independant Feature
X=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
from sklearn.model_selection import train_test_split
X_train,y_train,X_test,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lebal1=LabelEncoder()
lebal2=LabelEncoder()
lebal3=LabelEncoder()
```

```python
X_train['sex']=lebal1.fit_transform(X_train['sex'])
X_train['smoker']=lebal2.fit_transform(X_train['smoker'])
X_train['time']=lebal3.fit_transform(X_train['time'])
```
