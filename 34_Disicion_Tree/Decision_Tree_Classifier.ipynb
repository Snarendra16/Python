### Decision Tree Classifier

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

```python
# INDEPENDANT FEATURES
X=pd.DataFrame(iris['data'], columns=['sepal length in cm','sepal width in cm','petal length in cm','petal width in cm'])
```

```python
y=iris['target']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
DTC=DecisionTreeClassifier(max_depth=4)
# DTC=DecisionTreeClassifier()
```

```python
DTC.fit(X_train,y_train)
```

```python
# Visualize tree
from sklearn import tree
plt.Figure(figsize=(25,20))
tree.plot_tree(DTC,filled=True)
```

```python
# Prediction
y_pred=DTC.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

### Prepruning and HyperParameterTuning

```python
param={
    'criterion':['gini','entropy','log_loss'],
    'splitter' :['best','random'],
    'max_depth':[1,2,3,4,5,6,7],
    'max_features':['auto','sqrt','log2'],
}
```

```python
param
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid=GridSearchCV(DTC,param_grid=param,cv=5,scoring='accuracy')
```

```python
grid.fit(X_train,y_train)
```
