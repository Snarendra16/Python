```python
corpus="""Hey, Narendra here and I am Learning Natural Language Processing. I am loving it. It is so interesting and fun to learn. I am learning a lot of new things and I am enjoying it a lot. I am looking forward to learning more about Natural Language Processing and its applications in the real world."""
```

```python
corpus
```

```python

import nltk
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
tokens = word_tokenize(corpus)
tokens1=sent_tokenize(corpus)
print("Word tokenization")
print(tokens)
print("sentence tokenization")
print(tokens1)
```

```python
from nltk.tokenize import wordpunct_tokenize
tokens2=wordpunct_tokenize(corpus)
print("wordpunct_tokenize")
print(tokens2)


# Punctuation treated as separate tokens in wordpunct_tokenize but not in word_tokenize
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokens3=tokenizer.tokenize(corpus)
print("TreebankWordTokenizer")
print(tokens3)

# TreebankWordTokenizer is more sophisticated and handles contractions and punctuation better than word_tokenize and wordpunct_tokenize.
```
