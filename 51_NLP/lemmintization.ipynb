## Lammitization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
words=["running","ran","runs","easily","fairly"]
```

```python
for word in words:
    print(f"{word} --> {lemmatizer.lemmatize(word,pos='v')}")
```

pos = 1. verb (v)
      2. noun (n)
      3. adjective (a)
      4. adverb (r)
