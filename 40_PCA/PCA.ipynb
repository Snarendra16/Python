```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
```

```python
scalar.fit(df)
```

```python
scaled_data = scalar.transform(df)
```

```python
scaled_data
```

```python
# appliying PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
```

```python
pca.fit_transform(scaled_data)
```
